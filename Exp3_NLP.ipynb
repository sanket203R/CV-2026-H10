```python
# ============================================================
# 1. TOKENIZATION
# ============================================================

print("=" * 70)
print("1. TOKENIZATION")
print("=" * 70)

words = nltk.word_tokenize(resume_text)

print("\n--- Original Resume Text ---")
print(resume_text)

print("\n--- Word Tokens ---")
print(words)

print("\nTotal Number of Tokens:", len(words))

sentences = nltk.sent_tokenize(resume_text)

print("\n--- Sentence Tokens ---")

for i, sentence in enumerate(sentences, start=1):
    print(f"{i}. {sentence}")

print("\nTotal Number of Sentences:", len(sentences))



```

```python
# ============================================================
# 2. STEMMING
# ============================================================

print("\n" + "=" * 70)
print("2. STEMMING")
print("=" * 70)

porter = PorterStemmer()
lancaster = LancasterStemmer()
snowball = SnowballStemmer("english")

print("\n--- Stemming Results ---")

print(
    f"{'Original Word':20}"
    f"{'Porter':20}"
    f"{'Lancaster':20}"
    f"{'Snowball':20}"
)

print("-" * 80)

for word in words:

    if word.isalpha():

        porter_result = porter.stem(word)
        lancaster_result = lancaster.stem(word)
        snowball_result = snowball.stem(word)

        print(
            f"{word:20}"
            f"{porter_result:20}"
            f"{lancaster_result:20}"
            f"{snowball_result:20}"
        )


```

```python
# ============================================================
# 3. LEMMATIZATION
# ============================================================

print("\n" + "=" * 70)
print("3. LEMMATIZATION")
print("=" * 70)

lemmatizer = WordNetLemmatizer()

print("\n--- Lemmatization using WordNet ---")

print(
    f"{'Original Word':20}"
    f"{'Lemma':20}"
)

print("-" * 40)

for word in words:

    if word.isalpha():

        lemma = lemmatizer.lemmatize(word)

        print(
            f"{word:20}"
            f"{lemma:20}"
        )

```

```python
# ============================================================
# Lemmatization with POS Tags
# ============================================================

print("\n--- Lemmatization with POS Tags ---")

print(
    f"{'Original Word':20}"
    f"{'Noun':20}"
    f"{'Verb':20}"
)

print("-" * 60)

for word in words:

    if word.isalpha():

        noun_lemma = lemmatizer.lemmatize(
            word,
            pos='n'
        )

        verb_lemma = lemmatizer.lemmatize(
            word,
            pos='v'
        )

        print(
            f"{word:20}"
            f"{noun_lemma:20}"
            f"{verb_lemma:20}"
        )
```
